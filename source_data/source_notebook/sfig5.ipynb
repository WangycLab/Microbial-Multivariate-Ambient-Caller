```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from matplotlib.lines import Line2D

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 12

# ============================================================
# 1. Read
# ============================================================

file = "/Users/apple/Downloads/benchmarks/eco/test_555_rep1.barcodes.tsv"

df = pd.read_csv(file, sep='\t')

print(df.columns.tolist())

print("\nNumber of barcodes:")
print(len(df))

# ============================================================
# 2. Define calling groups from barcode format
#
# xxx#large  -> large
# xxx#small  -> small
# EMPTYxxxx  -> EMPTY
# ============================================================

barcode = df["barcode"].astype(str)

df["plot_group"] = "Other"

df.loc[barcode.str.endswith('#large'), 'plot_group'] = 'large'

df.loc[barcode.str.endswith('#small') | barcode.str.startswith('EMPTY'), 'plot_group'] = 'small_EMPTY'

print("\nPlot groups:")
print(df["plot_group"].value_counts())

# ============================================================
# 3. Sort by original rank
#
# The original rank may range from 2 to 6015.
# We use it only to determine the ordering of these 3000
# barcodes, then generate a new rank from 1 to 3000.
# ============================================================

plot_df = df[df['umis'] > 0].copy()

plot_df = plot_df.sort_values('rank').reset_index(drop=True)

plot_df['plot_rank'] = np.arange(1, len(plot_df) + 1)

print("\nNew plot rank:")
print("min =", plot_df["plot_rank"].min())
print("max =", plot_df["plot_rank"].max())

# ============================================================
# 4. Transform variables
# ============================================================

# y axis
plot_df['log_umi'] = np.log10(plot_df['umis'])

# x axis
#
# 0.5 = sqrt
# 0.75 = more expanded
# 1.0 = linear
#
plot_df['x_rank'] = plot_df['plot_rank'] ** 0.75

# ============================================================
# 5. LOWESS function
# ============================================================

def add_lowess(
    data,
    color="black",
    frac=0.025
):

    if len(data) < 10:
        return

    d = data.sort_values("x_rank")

    fit = lowess(
        endog=d["log_umi"].values,
        exog=d["x_rank"].values,
        frac=frac,
        it=3,
        return_sorted=True
    )

    ax.plot(
        fit[:, 0],
        fit[:, 1],
        color=color,
        linewidth=1,
        zorder=5
    )

# ============================================================
# 6. Create figure
# ============================================================

fig, ax = plt.subplots(figsize=(5.5, 4.5))

# ============================================================
# 7. Other barcodes
# ============================================================

other = plot_df[plot_df['plot_group'] == 'Other']

ax.scatter(other['x_rank'], other['log_umi'], s=14, alpha=0.2, color='0.75', linewidths=0, rasterized=True)

# ============================================================
# 8. Large cells
# ============================================================

large = plot_df[plot_df['plot_group'] == 'large']

ax.scatter(large['x_rank'], large['log_umi'], s=14, alpha=0.65, color='C0', linewidths=0, rasterized=True)

# ============================================================
# 9. Small + EMPTY
# ============================================================

small_empty = plot_df[plot_df['plot_group'] == 'small_EMPTY']

ax.scatter(
    small_empty["x_rank"],
    small_empty["log_umi"],
    s=14,
    alpha=0.65,
    color="#E9A3A3",
    linewidths=0,
    rasterized=True
)

# ============================================================
# 10. LOWESS
# ============================================================

add_lowess(large, color='black', frac=0.025)

add_lowess(small_empty, color='black', frac=0.025)

# ============================================================
# 11. X-axis ticks
# ============================================================

rank_ticks = np.array([1, 10, 100, 500, 1000, 1500, 2000, 2500, 3000])

rank_ticks = rank_ticks[rank_ticks <= plot_df['plot_rank'].max()]

ax.set_xticks(rank_ticks ** 0.75)

ax.set_xticklabels([str(x) for x in rank_ticks])

# ============================================================
# 12. Labels
# ============================================================

ax.set_xlabel('Barcode rank', fontsize=20)

ax.set_ylabel('$\\log_{10}$(UMIs)', fontsize=20)

# ============================================================
# 13. Style
# ============================================================

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.tick_params(direction='out', length=4, width=1, labelsize=15)

# ============================================================
# 14. Legend
# ============================================================

legend_handles = [
    Line2D(
        [0], [0],
        marker="o",
        color="C0",
        linestyle="None",
        markersize=5,
        label="Large cells"
    ),
    Line2D(
        [0], [0],
        marker="o",
        color="#E9A3A3",
        linestyle="None",
        markersize=5,
        label="Small cells / Empty mixture"
    ),
    Line2D(
        [0], [0],
        color="black",
        linewidth=1,
        label="Fitting"
    )
]

ax.legend(handles=legend_handles, frameon=False, loc='best', fontsize=15)

# ============================================================
# 15. Layout
# ============================================================

plt.tight_layout()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 10

# Nature-style axes / ticks
plt.rcParams["axes.linewidth"] = 0.8
plt.rcParams["axes.edgecolor"] = "black"
plt.rcParams["xtick.direction"] = "out"
plt.rcParams["ytick.direction"] = "out"
plt.rcParams["xtick.major.width"] = 0.8
plt.rcParams["ytick.major.width"] = 0.8
plt.rcParams["xtick.major.size"] = 3
plt.rcParams["ytick.major.size"] = 3
plt.rcParams["legend.frameon"] = False

# ============================================================
# Read the test files
#
# 3 replicates per subsampling scenario
# ============================================================

BASE = "/Users/apple/Downloads/benchmarks/eco"

FILES = ['test_555_rep1', 'test_555_rep2', 'test_555_rep3']

def kneedle_cut(umi):
    """Knee of the rank-UMI curve (linear UMI); returns the number
    of barcodes above the knee."""

    y = np.asarray(umi, dtype=float)
    x = np.arange(1, len(y) + 1)

    xn = (x - x.min()) / (x.max() - x.min())
    yn = (y - y.min()) / (y.max() - y.min())

    dx = xn[-1] - xn[0]
    dy = yn[-1] - yn[0]

    dist = np.abs(dx * (yn - yn[0]) - dy * (xn - xn[0])) / np.hypot(dx, dy)

    return int(np.argmax(dist)) + 1

rows = []

for rep in FILES:

    df = pd.read_csv(
        f"{BASE}/{rep}.barcodes.tsv",
        sep="\t",
        dtype={"barcode": str}
    )

    barcode = df["barcode"].astype(str)

    df["group"] = np.where(
        barcode.str.endswith("#large"), "large",
        np.where(
            barcode.str.endswith("#small"), "small",
            np.where(barcode.str.startswith("EMPTY"), "EMPTY", "other")
        )
    )

    # --------------------------------------------------------
    # Knee: top-N barcodes by UMI (no statistical test)
    # --------------------------------------------------------

    umi = pd.to_numeric(df["umis"], errors="coerce").fillna(0).values

    order = np.argsort(-umi, kind="stable")

    knee_mask = np.zeros(len(df), dtype=bool)

    knee_mask[order[:kneedle_cut(umi)]] = True

    # --------------------------------------------------------
    # Statistical methods
    # --------------------------------------------------------

    padj_total = pd.to_numeric(df["padj_total_rrna"], errors="coerce")
    padj_nonrrna = pd.to_numeric(df["padj_non_rrna"], errors="coerce")

    calls = {
        "MVAcall": df["call"].astype(str).str.lower().eq("cell").values,
        "Knee": knee_mask,
        "EmptyDrops": (padj_total < 0.01).values,
        "EmptyDrops\n(non-rRNA)": (padj_nonrrna < 0.01).values,
    }

    for method, mask in calls.items():

        counts = df.loc[mask, "group"].value_counts()

        rows.append({
            "method": method,
            "rep": rep,
            "large": int(counts.get("large", 0)),
            "small": int(counts.get("small", 0)),
            "empty": int(counts.get("EMPTY", 0)),
            "called": int(np.asarray(mask).sum()),
        })

res = pd.DataFrame(rows)

print(res.to_string(index=False))

# ============================================================
# Ground truth / denominators
# ============================================================

gt = df["group"].value_counts()

n_large = int(gt.get("large", 0))
n_small = int(gt.get("small", 0))
n_empty = int(gt.get("EMPTY", 0))

print(f"\nGround truth: large={n_large}, small={n_small}, background={n_empty}")

methods = ['MVAcall', 'Knee', 'EmptyDrops', 'EmptyDrops\n(non-rRNA)']

def get_counts(method, column):

    return np.array([
        res.loc[
            (res["method"] == method) & (res["rep"] == rep),
            column
        ].iloc[0]
        for rep in FILES
    ], dtype=float)

large = np.array([get_counts(m, "large") for m in methods])
small = np.array([get_counts(m, "small") for m in methods])
empty = np.array([get_counts(m, "empty") for m in methods])
called = np.array([get_counts(m, "called") for m in methods])

# Knee is a rank cutoff, not a statistical test -> no FDR
has_fdr = np.array([True, False, True, True])

# ============================================================
# Mean ± SEM
# ============================================================

large_mean = large.mean(axis=1) / n_large
large_sem = large.std(axis=1, ddof=1) / np.sqrt(large.shape[1]) / n_large

small_mean = small.mean(axis=1) / n_small
small_sem = small.std(axis=1, ddof=1) / np.sqrt(small.shape[1]) / n_small

# FDR = EMPTY called as cells / all barcodes called as cells
fdr_rep = np.where(called > 0, empty / called, np.nan)

fdr_mean = np.where(has_fdr, np.nanmean(fdr_rep, axis=1), np.nan)

fdr_sem = np.where(has_fdr, np.nanstd(fdr_rep, axis=1, ddof=1) / np.sqrt(fdr_rep.shape[1]), np.nan)

# ============================================================
# Plot parameters
# ============================================================

x = np.arange(len(methods)) * 0.62

bar_width = 0.09

bar_spacing = 0.10

# ============================================================
# Figure
# ============================================================

fig, ax1 = plt.subplots(figsize=(4, 5))

bars_large = ax1.bar(
    x - bar_spacing,
    large_mean,
    bar_width,
    yerr=large_sem,
    error_kw={
        "elinewidth": 0.7,
        "capsize": 2,
        "capthick": 0.7
    },
    edgecolor="black",
    linewidth=0.6,
    label="Large cells"
)

bars_small = ax1.bar(
    x,
    small_mean,
    bar_width,
    yerr=small_sem,
    error_kw={
        "elinewidth": 0.7,
        "capsize": 2,
        "capthick": 0.7
    },
    edgecolor="black",
    linewidth=0.6,
    label="Small cells"
)

ax1.set_ylabel("Recall")
# extra headroom when the Large / Small labels would collide
clash = np.abs(large_mean - small_mean) < 0.07

ax1.set_ylim(0, 1.22 if clash.any() else 1.12)

# ============================================================
# FDR: right axis (Knee is skipped: no FDR concept)
# ============================================================

ax2 = ax1.twinx()

bars_fdr = []

for i in range(len(methods)):

    if not has_fdr[i]:
        bars_fdr.append(None)
        continue

    bars_fdr.append(ax2.bar(
        x[i] + bar_spacing,
        fdr_mean[i],
        bar_width,
        yerr=fdr_sem[i],
        error_kw={
            "elinewidth": 0.8,
            "capsize": 2.5,
            "capthick": 0.8
        },
        color="red",
        edgecolor="black",
        linewidth=0.6,
        label="FDR" if i == 0 else None
    )[0])

ax2.set_ylabel("Observed FDR")

ax2.set_ylim(0, 0.04)

ax2.set_yticks([0, 0.01, 0.02, 0.03, 0.04])

ax1.spines["right"].set_visible(False)

ax2.spines["top"].set_visible(False)
ax2.spines["bottom"].set_visible(False)
ax2.spines["left"].set_visible(False)

ax1.tick_params(axis="both", labelsize=9, width=0.8, length=3, direction="out")
ax2.tick_params(axis="y", labelsize=9, width=0.8, length=3, direction="out")

# ============================================================
# Percentage labels
# ============================================================

def add_labels(ax, bars, values, offset, ha="center", dx=0.0):

    offsets = np.broadcast_to(offset, (len(values),))

    for bar, value, offset in zip(bars, values, offsets):

        if bar is None or not np.isfinite(value):
            continue

        height = bar.get_height()

        if value == 0:
            label = "0%"
        else:
            label = f"{value * 100:.1f}%"

        # zero-height bars: shift the label right so it does not
        # sit on top of the neighbouring bar
        if value == 0:
            # zero-height bars: place the label outside the bar pair
            if ha == "center":
                ax.text(
                    bar.get_x() - 0.005,
                    height + offset,
                    label,
                    ha="right",
                    va="bottom",
                    fontsize=8
                )
            else:
                ax.text(
                    bar.get_x() + bar.get_width() + 0.005,
                    height + offset,
                    label,
                    ha="left",
                    va="bottom",
                    fontsize=8
                )
        else:
            ax.text(
                bar.get_x() + bar.get_width() / 2 + dx,
                height + offset,
                label,
                ha=ha,
                va="bottom",
                fontsize=8
            )

add_labels(ax1, bars_large, large_mean, offset=0.05)
add_labels(
    ax1,
    bars_small,
    small_mean,
    offset=np.where(clash, large_mean + 0.115 - small_mean, 0.05),
    ha="left",
    dx=0.055
)
add_labels(ax2, bars_fdr, fdr_mean, offset=0.0012, ha="left", dx=0.012)

# ============================================================
# X axis
# ============================================================

ax1.set_xlim(x[0] - 0.30, x[-1] + 0.42)

ax1.set_xticks(x)

ax1.set_xticklabels(methods, rotation=30, ha='right', fontsize=9)

ax1.set_title(f'Large/Small/Background {n_large}/{n_small}/{n_empty}', fontsize=10, pad=8)

# ============================================================
# Combined legend
# ============================================================

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc="upper right",
    fontsize=9,
    handlelength=1.2,
    handletextpad=0.5,
    labelspacing=0.35
)

# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()

print("\nmean recall large:", np.round(large_mean, 4))
print("mean recall small:", np.round(small_mean, 4))
print("mean FDR        :", np.round(fdr_mean, 4))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 10

# Nature-style axes / ticks
plt.rcParams["axes.linewidth"] = 0.8
plt.rcParams["axes.edgecolor"] = "black"
plt.rcParams["xtick.direction"] = "out"
plt.rcParams["ytick.direction"] = "out"
plt.rcParams["xtick.major.width"] = 0.8
plt.rcParams["ytick.major.width"] = 0.8
plt.rcParams["xtick.major.size"] = 3
plt.rcParams["ytick.major.size"] = 3
plt.rcParams["legend.frameon"] = False

# ============================================================
# Read the test files
#
# 3 replicates per subsampling scenario
# ============================================================

BASE = "/Users/apple/Downloads/benchmarks/eco"

FILES = ['test_5125_rep1', 'test_5125_rep2', 'test_5125_rep3']

def kneedle_cut(umi):
    """Knee of the rank-UMI curve (linear UMI); returns the number
    of barcodes above the knee."""

    y = np.asarray(umi, dtype=float)
    x = np.arange(1, len(y) + 1)

    xn = (x - x.min()) / (x.max() - x.min())
    yn = (y - y.min()) / (y.max() - y.min())

    dx = xn[-1] - xn[0]
    dy = yn[-1] - yn[0]

    dist = np.abs(dx * (yn - yn[0]) - dy * (xn - xn[0])) / np.hypot(dx, dy)

    return int(np.argmax(dist)) + 1

rows = []

for rep in FILES:

    df = pd.read_csv(
        f"{BASE}/{rep}.barcodes.tsv",
        sep="\t",
        dtype={"barcode": str}
    )

    barcode = df["barcode"].astype(str)

    df["group"] = np.where(
        barcode.str.endswith("#large"), "large",
        np.where(
            barcode.str.endswith("#small"), "small",
            np.where(barcode.str.startswith("EMPTY"), "EMPTY", "other")
        )
    )

    # --------------------------------------------------------
    # Knee: top-N barcodes by UMI (no statistical test)
    # --------------------------------------------------------

    umi = pd.to_numeric(df["umis"], errors="coerce").fillna(0).values

    order = np.argsort(-umi, kind="stable")

    knee_mask = np.zeros(len(df), dtype=bool)

    knee_mask[order[:kneedle_cut(umi)]] = True

    # --------------------------------------------------------
    # Statistical methods
    # --------------------------------------------------------

    padj_total = pd.to_numeric(df["padj_total_rrna"], errors="coerce")
    padj_nonrrna = pd.to_numeric(df["padj_non_rrna"], errors="coerce")

    calls = {
        "MVAcall": df["call"].astype(str).str.lower().eq("cell").values,
        "Knee": knee_mask,
        "EmptyDrops": (padj_total < 0.01).values,
        "EmptyDrops\n(non-rRNA)": (padj_nonrrna < 0.01).values,
    }

    for method, mask in calls.items():

        counts = df.loc[mask, "group"].value_counts()

        rows.append({
            "method": method,
            "rep": rep,
            "large": int(counts.get("large", 0)),
            "small": int(counts.get("small", 0)),
            "empty": int(counts.get("EMPTY", 0)),
            "called": int(np.asarray(mask).sum()),
        })

res = pd.DataFrame(rows)

print(res.to_string(index=False))

# ============================================================
# Ground truth / denominators
# ============================================================

gt = df["group"].value_counts()

n_large = int(gt.get("large", 0))
n_small = int(gt.get("small", 0))
n_empty = int(gt.get("EMPTY", 0))

print(f"\nGround truth: large={n_large}, small={n_small}, background={n_empty}")

methods = ['MVAcall', 'Knee', 'EmptyDrops', 'EmptyDrops\n(non-rRNA)']

def get_counts(method, column):

    return np.array([
        res.loc[
            (res["method"] == method) & (res["rep"] == rep),
            column
        ].iloc[0]
        for rep in FILES
    ], dtype=float)

large = np.array([get_counts(m, "large") for m in methods])
small = np.array([get_counts(m, "small") for m in methods])
empty = np.array([get_counts(m, "empty") for m in methods])
called = np.array([get_counts(m, "called") for m in methods])

# Knee is a rank cutoff, not a statistical test -> no FDR
has_fdr = np.array([True, False, True, True])

# ============================================================
# Mean ± SEM
# ============================================================

large_mean = large.mean(axis=1) / n_large
large_sem = large.std(axis=1, ddof=1) / np.sqrt(large.shape[1]) / n_large

small_mean = small.mean(axis=1) / n_small
small_sem = small.std(axis=1, ddof=1) / np.sqrt(small.shape[1]) / n_small

# FDR = EMPTY called as cells / all barcodes called as cells
fdr_rep = np.where(called > 0, empty / called, np.nan)

fdr_mean = np.where(has_fdr, np.nanmean(fdr_rep, axis=1), np.nan)

fdr_sem = np.where(has_fdr, np.nanstd(fdr_rep, axis=1, ddof=1) / np.sqrt(fdr_rep.shape[1]), np.nan)

# ============================================================
# Plot parameters
# ============================================================

x = np.arange(len(methods)) * 0.62

bar_width = 0.09

bar_spacing = 0.10

# ============================================================
# Figure
# ============================================================

fig, ax1 = plt.subplots(figsize=(4, 5))

bars_large = ax1.bar(
    x - bar_spacing,
    large_mean,
    bar_width,
    yerr=large_sem,
    error_kw={
        "elinewidth": 0.7,
        "capsize": 2,
        "capthick": 0.7
    },
    edgecolor="black",
    linewidth=0.6,
    label="Large cells"
)

bars_small = ax1.bar(
    x,
    small_mean,
    bar_width,
    yerr=small_sem,
    error_kw={
        "elinewidth": 0.7,
        "capsize": 2,
        "capthick": 0.7
    },
    edgecolor="black",
    linewidth=0.6,
    label="Small cells"
)

ax1.set_ylabel("Recall")
# extra headroom when the Large / Small labels would collide
clash = np.abs(large_mean - small_mean) < 0.07

ax1.set_ylim(0, 1.22 if clash.any() else 1.12)

# ============================================================
# FDR: right axis (Knee is skipped: no FDR concept)
# ============================================================

ax2 = ax1.twinx()

bars_fdr = []

for i in range(len(methods)):

    if not has_fdr[i]:
        bars_fdr.append(None)
        continue

    bars_fdr.append(ax2.bar(
        x[i] + bar_spacing,
        fdr_mean[i],
        bar_width,
        yerr=fdr_sem[i],
        error_kw={
            "elinewidth": 0.8,
            "capsize": 2.5,
            "capthick": 0.8
        },
        color="red",
        edgecolor="black",
        linewidth=0.6,
        label="FDR" if i == 0 else None
    )[0])

ax2.set_ylabel("Observed FDR")

ax2.set_ylim(0, 0.04)

ax2.set_yticks([0, 0.01, 0.02, 0.03, 0.04])

ax1.spines["right"].set_visible(False)

ax2.spines["top"].set_visible(False)
ax2.spines["bottom"].set_visible(False)
ax2.spines["left"].set_visible(False)

ax1.tick_params(axis="both", labelsize=9, width=0.8, length=3, direction="out")
ax2.tick_params(axis="y", labelsize=9, width=0.8, length=3, direction="out")

# ============================================================
# Percentage labels
# ============================================================

def add_labels(ax, bars, values, offset, ha="center", dx=0.0):

    offsets = np.broadcast_to(offset, (len(values),))

    for bar, value, offset in zip(bars, values, offsets):

        if bar is None or not np.isfinite(value):
            continue

        height = bar.get_height()

        if value == 0:
            label = "0%"
        else:
            label = f"{value * 100:.1f}%"

        # zero-height bars: shift the label right so it does not
        # sit on top of the neighbouring bar
        if value == 0:
            # zero-height bars: place the label outside the bar pair
            if ha == "center":
                ax.text(
                    bar.get_x() - 0.005,
                    height + offset,
                    label,
                    ha="right",
                    va="bottom",
                    fontsize=8
                )
            else:
                ax.text(
                    bar.get_x() + bar.get_width() + 0.005,
                    height + offset,
                    label,
                    ha="left",
                    va="bottom",
                    fontsize=8
                )
        else:
            ax.text(
                bar.get_x() + bar.get_width() / 2 + dx,
                height + offset,
                label,
                ha=ha,
                va="bottom",
                fontsize=8
            )

add_labels(ax1, bars_large, large_mean, offset=0.05)
add_labels(
    ax1,
    bars_small,
    small_mean,
    offset=np.where(clash, large_mean + 0.115 - small_mean, 0.05),
    ha="left",
    dx=0.055
)
add_labels(ax2, bars_fdr, fdr_mean, offset=0.0012, ha="left", dx=0.012)

# ============================================================
# X axis
# ============================================================

ax1.set_xlim(x[0] - 0.30, x[-1] + 0.42)

ax1.set_xticks(x)

ax1.set_xticklabels(methods, rotation=30, ha='right', fontsize=9)

ax1.set_title(f'Large/Small/Background {n_large}/{n_small}/{n_empty}', fontsize=10, pad=8)

# ============================================================
# Combined legend
# ============================================================

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc="upper right",
    fontsize=9,
    handlelength=1.2,
    handletextpad=0.5,
    labelspacing=0.35
)

# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()

print("\nmean recall large:", np.round(large_mean, 4))
print("mean recall small:", np.round(small_mean, 4))
print("mean FDR        :", np.round(fdr_mean, 4))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 12

# ==========================
# 读取原始 barcode 数据
# ==========================
df = pd.read_csv('/Users/apple/Downloads/benchmarks/eco/run1.barcodes.tsv', sep='\t')

# ==========================
# Method 1: call == cell
# ==========================
df_cells_method1 = df[df['call'] == 'cell'].copy()

# ==========================
# Method 2: Kneedle
# ==========================
df_sorted = df.sort_values('rank').reset_index(drop=True)

df_cells_method2 = df_sorted[df_sorted.index < 1871].copy()

# ==========================
# Method 3: padj_non_rrna < 0.01
# ==========================
df_cells_method3 = df[df['padj_non_rrna'].notna() & (df['padj_non_rrna'] < 0.01)].copy()

# ==========================
# Method 4: padj_gene < 0.01
# 使用另一个文件的barcode筛选
# ==========================

df_gene = pd.read_csv('/Users/apple/Downloads/benchmarks/eco/total1.barcodes.tsv', sep='\t')

method4_barcodes = df_gene.loc[
    (df_gene['padj_non_rrna'].notna()) &
    (df_gene['padj_non_rrna'] < 0.01),
    'barcode'
]

# 从原始df取数据，保证UMI一致
df_cells_method4 = df[df['barcode'].isin(method4_barcodes)].copy()

# ==========================
# 创建 UMI bins
# ==========================

min_umis = min(
    df_cells_method1['umis'].min(),
    df_cells_method2['umis'].min(),
    df_cells_method3['umis'].min(),
    df_cells_method4['umis'].min()
)

max_umis = max(
    df_cells_method1['umis'].max(),
    df_cells_method2['umis'].max(),
    df_cells_method3['umis'].max(),
    df_cells_method4['umis'].max()
)

bins = np.logspace(np.log10(min_umis), np.log10(max_umis), 30)

# ==========================
# 统计每个method的UMI bin分布
# ==========================

def get_bin_stats(data):

    data = data.copy()

    data['log_umi'] = np.log10(data['umis'])

    data['umi_bin'] = pd.cut(
        data['umis'],
        bins=bins,
        include_lowest=True
    )

    stats = data.groupby(
        'umi_bin',
        observed=False
    ).agg(
        cell_count=('barcode','count')
    ).reset_index()

    return stats

bin_stats1 = get_bin_stats(df_cells_method1)

bin_stats2 = get_bin_stats(df_cells_method2)

bin_stats3 = get_bin_stats(df_cells_method3)

bin_stats4 = get_bin_stats(df_cells_method4)

# ==========================
# x轴(log10 bin)
# ==========================

x_step = np.log10(bins)[:-1]

# ==========================
# 绘图
# ==========================

fig, ax = plt.subplots(figsize=(8, 6))

# Method 1
y1 = bin_stats1['cell_count'].values

ax.step(x_step, y1, where='pre', linewidth=1, color='#2E86C1', label='ModuleDrops')

# Method 2
y2 = bin_stats2['cell_count'].values

ax.step(x_step, y2, where='pre', linewidth=1, color='#E74C3C', label='Knee')

# Method 3
y3 = bin_stats3['cell_count'].values

ax.step(x_step, y3, where='pre', linewidth=1, color='#28B463', label='EmptyDrops non-rRNA')

# Method 4
y4 = bin_stats4['cell_count'].values

ax.step(x_step, y4, where='pre', linewidth=1, color='#8E44AD', label='EmptyDrops total RNA')

# ==========================
# 图形设置
# ==========================

ax.set_xlabel('log10(UMI per barcode)', fontsize=20)

ax.set_ylabel('Number of Cells', fontsize=20)

ax.set_xlim(x_step.min(), x_step.max())

# legend放最后
ax.legend(fontsize=16)

# ==========================
# 统计信息
# ==========================

methods = [
    ('ModuleDrops', df_cells_method1),
    ('Knee', df_cells_method2),
    ('EmptyDrops non-rRNA', df_cells_method3),
    ('EmptyDrops total RNA', df_cells_method4)
]

ax.tick_params(labelsize=16)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 12

# ============================================================
# 1. Read counts.tsv
# ============================================================

counts_file = '/Users/apple/Downloads/benchmarks/eco/counts.tsv'

df = pd.read_csv(counts_file, sep='\t', dtype={'barcode': str})

print("counts.tsv columns:")
print(df.columns.tolist())

print(f'\n总 barcode 数: {len(df)}')
print(f'总 UMI 数: {df["umi_count"].sum():,}')
print(f"UMI 范围: {df['umi_count'].min()} ~ {df['umi_count'].max()}")

# ============================================================
# 2. Sort by UMI count and assign rank
# ============================================================

df_sorted = df.sort_values('umi_count', ascending=False).reset_index(drop=True)

df_sorted['rank'] = np.arange(1, len(df_sorted) + 1)

# 确保 barcode 是字符串
df_sorted['barcode'] = df_sorted['barcode'].astype(str)

# ============================================================
# 3. Read run1.barcodes.tsv
# ============================================================

barcode_file = '/Users/apple/Downloads/benchmarks/eco/run1.barcodes.tsv'

barcode_df = pd.read_csv(barcode_file, sep='\t', dtype={'barcode': str})

barcode_df['barcode'] = barcode_df['barcode'].astype(str)

print(f'\nrun1 barcode 数: {len(barcode_df)}')

print(f"cell barcode 数: {(barcode_df['call'] == 'cell').sum()}")

# ============================================================
# 4. Check barcode overlap
# ============================================================

counts_barcodes = set(df_sorted['barcode'])

run1_barcodes = set(barcode_df['barcode'])

overlap = counts_barcodes & run1_barcodes

print(f'\nbarcode overlap: {len(overlap)}')

# ============================================================
# 5. Identify cell barcodes
# ============================================================

cell_barcodes = set(barcode_df.loc[barcode_df['call'] == 'cell', 'barcode'])

# 在 counts.tsv 中标记 cell
df_sorted['is_cell'] = df_sorted['barcode'].isin(cell_barcodes)

print(f"匹配到的 cell barcode 数: {df_sorted['is_cell'].sum()}")

# ============================================================
# 6. Define log-spaced bins
# ============================================================

max_rank = len(df_sorted)

n_bins = 30

bins = np.logspace(np.log10(1), np.log10(max_rank + 1), n_bins + 1)

# ============================================================
# 7. Assign run1 barcodes to rank bins
# ============================================================

barcode_df['rank_bin'] = pd.cut(barcode_df['rank'], bins=bins, include_lowest=True, right=False)

# ============================================================
# 8. Calculate cell percentage per bin
# ============================================================

bin_pct = (
    barcode_df
    .groupby(
        'rank_bin',
        observed=True
    )
    .agg(
        total=('call', 'size'),
        cells=(
            'call',
            lambda x: (
                x == 'cell'
            ).sum()
        )
    )
)

bin_pct['cell_pct'] = bin_pct['cells'] / bin_pct['total'] * 100

# ============================================================
# 9. Exact bin boundaries
# ============================================================

bin_left = np.array([interval.left for interval in bin_pct.index])

bin_right = np.array([interval.right for interval in bin_pct.index])

# log scale 下的几何中心
bin_center = np.sqrt(bin_left * bin_right)

# 真实 bin 宽度
bin_width = bin_right - bin_left

# ============================================================
# 10. Plot
# ============================================================

fig, ax = plt.subplots(figsize=(7, 6))

# ============================================================
# Colors
# ============================================================

grey = '0.65'
blue = 'tab:blue'

# ============================================================
# 11. All barcode scatter — grey
# ============================================================

ax.scatter(
    df_sorted['rank'],
    df_sorted['umi_count'],
    s=16,
    color=grey,
    edgecolors='none',
    zorder=2,
    rasterized=True
)

# ============================================================
# 12. Cell barcode scatter — blue
# ============================================================

cell_points = df_sorted[df_sorted['is_cell']]

ax.scatter(
    cell_points['rank'],
    cell_points['umi_count'],
    s=10,
    color=blue,
    edgecolors='none',
    zorder=3,
    rasterized=True
)

# ============================================================
# 13. Left axis
# ============================================================

ax.set_xlabel('Barcode Rank', fontsize=20)

ax.set_ylabel('Total UMI Count', fontsize=20)

ax.set_xscale('log')
ax.set_yscale('log')

# ============================================================
# 14. Right axis
# ============================================================

ax2 = ax.twinx()

# ============================================================
# 15. Cell percentage bin chart
# ============================================================

ax2.bar(
    bin_center,
    bin_pct['cell_pct'],
    width=bin_width,
    align='center',
    color=blue,
    alpha=0.35,
    edgecolor='none',
    linewidth=0,
    zorder=1
)

# ============================================================
# 16. Right axis settings
# ============================================================

ax2.set_ylabel('cell called(%)', fontsize=20)

ax2.set_ylim(0, 105)

ax2.set_xscale('log')

ax2.set_xlim(ax.get_xlim())

# ============================================================
# 17. Layout
# ============================================================

ax.tick_params(labelsize=16)
ax2.tick_params(labelsize=16)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica", "DejaVu Sans"]
plt.rcParams["font.size"] = 12

# ============================================================
# 1. Read data
# ============================================================

file = "/Users/apple/Downloads/benchmarks/eco/test_555_rep1.barcodes.tsv"

df = pd.read_csv(file, sep='\t')

print("Total barcodes:", len(df))

# ============================================================
# 2. Keep only call == cell
# ============================================================

df = df[df['call'].astype(str).str.lower().eq('cell')].copy()

print("Cell barcodes:", len(df))

# ============================================================
# 3. P-value columns
# ============================================================

pcols = {'rRNA%': 'pvalue_rrnapct', 'KM': 'pvalue_ko', 'non-rRNA': 'pvalue_non_rrna'}

for col in pcols.values():

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    df[col] = df[col].clip(
        lower=0,
        upper=1
    )

# ============================================================
# 4. Convert P to -log10(P)
# ============================================================

eps = np.finfo(float).tiny

for name, col in pcols.items():

    df[f"logp_{name}"] = -np.log10(
        df[col].clip(lower=eps)
    )

# ============================================================
# 5. Pairwise combinations
# ============================================================

pairs = [
    (
        "rRNA%",
        "KM",
        "rRNA% vs KM"
    ),
    (
        "rRNA%",
        "non-rRNA",
        "rRNA% vs non-rRNA"
    ),
    (
        "KM",
        "non-rRNA",
        "KM vs non-rRNA"
    )
]

# ============================================================
# 6. Figure
# ============================================================

fig = plt.figure(figsize=(12, 3.5))

gs = fig.add_gridspec(1, 4, width_ratios=[1, 1, 1, 0.06], wspace=0.35)

axes = [fig.add_subplot(gs[0, 0]), fig.add_subplot(gs[0, 1]), fig.add_subplot(gs[0, 2])]

cax = fig.add_subplot(gs[0, 3])

# ============================================================
# Plot
# ============================================================

hb = None

for ax, (xname, yname, title) in zip(
    axes,
    pairs
):

    x = df[f"logp_{xname}"].to_numpy()
    y = df[f"logp_{yname}"].to_numpy()

    valid = (
        np.isfinite(x) &
        np.isfinite(y)
    )

    x = x[valid]
    y = y[valid]

    rho, p_corr = spearmanr(x, y)

    hb = ax.hexbin(
        x,
        y,
        gridsize=45,
        mincnt=1,
        bins="log",
        linewidths=0.15
    )

    threshold = 2

    ax.axvline(
        threshold,
        linestyle="--",
        linewidth=0.8
    )

    ax.axhline(
        threshold,
        linestyle="--",
        linewidth=0.8
    )

    ax.set_xlabel(
        f"$-\\log_{{10}}(P)$ — {xname}",
        fontsize=11
    )

    ax.set_ylabel(
        f"$-\\log_{{10}}(P)$ — {yname}",
        fontsize=11
    )

    ax.set_title(
        title,
        fontsize=12
    )

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.tick_params(
        labelsize=9
    )

# ============================================================
# Shared colorbar
# ============================================================

cbar = fig.colorbar(hb, cax=cax)

cbar.set_label('Barcode count per hexbin', fontsize=10)

cbar.ax.tick_params(labelsize=9)

plt.show()

# ============================================================
# Layout
# ============================================================

fig.suptitle('Relationship between independent tests', fontsize=13, y=1.02)

fig.subplots_adjust(left=0.07, right=0.96, bottom=0.16, top=0.88, wspace=0.35)

plt.show()
plt.close(fig)
```
